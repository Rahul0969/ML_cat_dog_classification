```python
#pandas
import os
import glob
import pandas as pd

TRAIN_DIR = "C:/Users/vijay/Downloads/New folder/Machine-Learning/data/train"
TEST_DIR = "C:/Users/vijay/Downloads/New folder/Machine-Learning/data/test"

image_paths = glob.glob(os.path.join(TRAIN_DIR,"*.jpg"))
print("Total training images:",len(image_paths))

print("\n Sampled Images:")
for path in image_paths[:10]:
    print(path)
```

```python
import numpy as np
from pathlib import Path

TRAIN_DIR = Path("../data/train")

data = []

for image_path in TRAIN_DIR.iterdir():
    filename = os.path.basename(image_path).lower()

    if filename.startswith("cat"):
        label = 0
    elif filename.startswith("dog"):
        label = 1
    else:
        continue
    data.append({
        "filepath": image_path,
        "label" : label
    })

df = pd.DataFrame(data)

print("\nTotal images",len(df))
print(df.head())

print("\nClass distribution:")
print(df["label"].value_counts())
        
```

```python
from PIL import Image
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))

for i in range(8):
    image = Image.open(df.iloc[i]["filepath"])
    plt.subplot(2,4,i+1)
    plt.imshow(image)

    plt.title("Cat" if df.iloc[i]["label"] == 0 else "Dog")
    plt.axis("off")

plt.tight_layout()
plt.show()

```
